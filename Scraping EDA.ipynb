```python
from apify_client import ApifyClient
import pandas as pd

# Initialize the ApifyClient with your API token
client = ApifyClient("Apify_API_Token")  # Ganti dengan token API Anda

# Prepare the Actor input (Disesuaikan dengan masa jabatan Menkeu Purbaya: 8 Sep 2025 - 14 Sep 2026)
run_input = {
    "filter:blue_verified": False,
    "filter:consumer_video": False,
    "filter:has_engagement": False,
    "filter:hashtags": False,
    "filter:images": False,
    "filter:links": False,
    "filter:media": False,
    "filter:mentions": False,
    "filter:native_video": False,
    "filter:nativeretweets": False,
    "filter:news": False,
    "filter:pro_video": False,
    "filter:quote": False,
    "filter:replies": False,
    "filter:safe": False,
    "filter:spaces": False,
    "filter:twimg": False,
    "filter:videos": False,
    "filter:vine": False,
    "include:nativeretweets": False,
    "lang": "in",
    "maxItems": 10000,
    "start": "2025-09-08",
    "end": "2026-09-14",
    "searchTerms": [
        "(purbaya OR menkeu OR kemenkeu) (fiskal OR apbn OR pajak OR himbara) since:2025-09-08 until:2026-09-14",
        "\"kebijakan fiskal\" (purbaya OR menkeu OR prabowo) since:2025-09-08 until:2026-09-14",
        "\"menteri keuangan\" purbaya since:2025-09-08 until:2026-09-14",
        "apbn purbaya since:2025-09-08 until:2026-09-14",
        "pajak purbaya since:2025-09-08 until:2026-09-14",
        "himbara purbaya since:2025-09-08 until:2026-09-14"
    ]
}

# Run the Actor and wait for it to finish
run = client.actor("CJdippxWmn9uRfooo").call(run_input=run_input)

# Fetch and print Actor results from the run's dataset (if there are any)
items = []
dataset_id = run.default_dataset_id

for item in client.dataset(dataset_id).iterate_items():
    items.append(item)

# Penanganan jika data berhasil diambil
if len(items) > 0:
    df = pd.DataFrame(items)
    # Hapus duplikat jika ada cuitan yang terjaring oleh lebih dari satu searchTerm
    if 'id' in df.columns:
        df = df.drop_duplicates(subset=['id']).reset_index(drop=True)

    try:
        # Mencoba menyimpan ke format Excel
        df.to_excel('DataTwitter_Purbaya_Fiskal.xlsx', index=False)
        print(f"Berhasil mengambil {len(df)} data unik dan menyimpannya ke 'DataTwitter_Purbaya_Fiskal.xlsx'")
    except Exception as excel_err:
        print(f"Gagal menyimpan ke format Excel karena ketidakcocokan sistem: {excel_err}")
        # Fallback instan ke format CSV jika Excel mengalami kegagalan engine
        df.to_csv('DataTwitter_Purbaya_Fiskal.csv', index=False)
        print("Data berhasil disimpan ke format alternatif: 'DataTwitter_Purbaya_Fiskal.csv'")
else:
    print("Peringatan: Tidak ada data yang ditemukan dari scraping Apify (0 items).")
    df = pd.DataFrame(columns=["id", "text", "created_at", "author_username"])

display(df.head())
```
